```python

import pandas as pd
import matplotlib.pyplot as plt
import warnings 
warnings.filterwarnings('ignore')
import seaborn as sns
import numpy as np

# Load the dataset
file_path = 'Daily_Activity_metrics_filled5.csv'
data = pd.read_csv(file_path)

# # Convert 'Date' column to datetime format for consistency in time-based analysis
# data['Date'] = pd.to_datetime(data['Date'], format='%d-%m-%Y')
    
```

```python
data.columns
```

```python
data['Calculated Calories'] = (
    data['Move Minutes count'] * 0.15 +
    data['Calories (kcal)'] * 0.45 +
    data['Distance (m)'] * 0.05 +
    data['Average heart rate (bpm)'] * 0.1 +
    data['Max heart rate (bpm)'] * 0.05 +
    data['Min heart rate (bpm)'] * 0.05 +
    data['Average speed (m/s)'] * 0.3 +
    data['Max speed (m/s)'] * 0.15 +
    data['Min speed (m/s)'] * 0.05 +
    data['Step count'] * 0.05 +
    data['Walking duration (ms)'] * 0.000005
)*1.5
```

```python
data["Calories (kcal)"]=data["Calculated Calories"]
```

```python
data.drop(['Calculated Calories'], axis=1, inplace=True)
```
